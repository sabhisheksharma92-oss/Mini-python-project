## Project overview
This notebook analyses basic fitness information for a small group of people. It calculates Body Mass Index (BMI), assigns a simple BMI category, estimates daily calorie needs using a beginner-level formula, and displays a summary using Pandas and NumPy.

> **Educational note:** This is a classroom demonstration, not a medical tool. BMI and calorie estimates are rough screening estimates and should not be used to diagnose a health condition or prescribe a diet.


## 1. Import the required libraries
We use:
- **Pandas** to store and analyse tabular data.
- **NumPy** for numerical calculations.
- **Python** for functions, conditions and user input.


```python
import pandas as pd
import numpy as np
```

## 2. Create a sample dataset
The sample records make it possible to run the project immediately. You can edit the values or add more people.

**Units:** height in centimetres, weight in kilograms, age in years.


```python
people = {
    "Name": ["Aarav", "Diya", "Rohan", "Meera", "Kabir"],
    "Age": [19, 20, 21, 18, 22],
    "Gender": ["Male", "Female", "Male", "Female", "Male"],
    "Height_cm": [172, 160, 178, 155, 180],
    "Weight_kg": [68, 55, 82, 48, 75],
    "Activity_Level": ["Moderate", "Light", "High", "Light", "Moderate"]
}

df = pd.DataFrame(people)
df
```

## 3. Define calculation functions
BMI = weight (kg) / height (m)².

For a simple calorie estimate, this project uses the Mifflin-St Jeor equation for estimated resting energy and multiplies it by an activity factor. This is an approximation for adults.


```python
def calculate_bmi(weight_kg, height_cm):
    height_m = height_cm / 100
    return weight_kg / (height_m ** 2)


def bmi_category(bmi):
    if bmi < 18.5:
        return "Underweight"
    elif bmi < 25:
        return "Normal range"
    elif bmi < 30:
        return "Overweight"
    else:
        return "Obesity range"


def estimate_daily_calories(age, gender, height_cm, weight_kg, activity_level):
    # Mifflin-St Jeor resting-energy estimate
    if gender.lower() == "male":
        bmr = (10 * weight_kg) + (6.25 * height_cm) - (5 * age) + 5
    else:
        bmr = (10 * weight_kg) + (6.25 * height_cm) - (5 * age) - 161

    activity_factors = {
        "light": 1.375,
        "moderate": 1.55,
        "high": 1.725
    }
    factor = activity_factors.get(activity_level.lower(), 1.2)
    return bmr * factor
```

## 4. Analyse the dataset
The `apply()` method runs the functions for each row. NumPy is used to round the calculated values.


```python
df["BMI"] = df.apply(
    lambda row: calculate_bmi(row["Weight_kg"], row["Height_cm"]), axis=1
)
df["BMI"] = np.round(df["BMI"], 2)

df["BMI_Category"] = df["BMI"].apply(bmi_category)

df["Estimated_Daily_Calories"] = df.apply(
    lambda row: estimate_daily_calories(
        row["Age"], row["Gender"], row["Height_cm"],
        row["Weight_kg"], row["Activity_Level"]
    ),
    axis=1
)
df["Estimated_Daily_Calories"] = np.round(df["Estimated_Daily_Calories"]).astype(int)

df
```

## 5. Display useful summaries


```python
print("Total people analysed:", len(df))
print("\nBMI category count:")
print(df["BMI_Category"].value_counts())

print("\nAverage BMI:", round(df["BMI"].mean(), 2))
print("Average weight (kg):", round(df["Weight_kg"].mean(), 2))
print("Average estimated daily calories:", round(df["Estimated_Daily_Calories"].mean()))
```

## 6. Add your own person's details
Run this cell and enter the requested details when prompted. Activity level must be `Light`, `Moderate`, or `High`.


```python
name = input("Enter name: ")
age = int(input("Enter age in years: "))
gender = input("Enter gender (Male/Female): ")
height = float(input("Enter height in cm: "))
weight = float(input("Enter weight in kg: "))
activity = input("Enter activity level (Light/Moderate/High): ")

bmi = calculate_bmi(weight, height)
calories = estimate_daily_calories(age, gender, height, weight, activity)

print("\n--- Fitness Analysis ---")
print("Name:", name)
print("BMI:", round(bmi, 2))
print("BMI category:", bmi_category(bmi))
print("Estimated daily calories:", round(calories), "kcal/day")
```
